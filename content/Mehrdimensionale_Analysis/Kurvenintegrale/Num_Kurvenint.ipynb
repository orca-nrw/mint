### I. Kurvenlänge
<hr>

Möchte man die Fläche berechnen, die zwischen dem Graphen einer gegebenen Funktion $f(x)$ und der $x$-Achse von $x=a$ bis $x=b$ eingeschlossen wird, so kann man die Fläche zunächst als eine Reihe von Rechtecken gleicher Breite $\Delta x$ approximieren und über die Flächen aller Rechtecke summieren. Die Höhe der Rechtecke ist dabei jeweils durch den Funktionswert an einem der beiden Eckpunkte gegeben; hier dem linken, wie in der unteren, linken Skizze gezeigt. Verkleinert man nun die Breite der Rechtecke $\Delta x$ und erhöht damit die Anzahl dieser, nähert man sich der exakten Fläche unter dem Graphen immer weiter an. Diese Herangehensweise ist Ihnen wahrscheinlich schon aus der Schule bekannt (Riemannsumme zur Flächenberechnung). 

Im nächsten Schritt möchten wir, anstatt der Fläche unterhalb des Graphens, die Länge des Graphens selbst berechnen. Auch hier werden wir wieder mit einer Approximation beginnen und uns dem exakten Wert anschließend beliebig genau annähern. Der Graph kann durch eine Reihe von Geradenstücken approxmiert werden, wie in der unteren, rechten Skizze dargestellt ist. Die Projektion jedes Geradenstücks auf die $x$-Achse soll dabei $\Delta x$ sein. Nach Pythagoras ist die Länge des jeweiligen Geradenstücks dann $\Delta s_j = \sqrt{(\Delta x)^2 + (\Delta y_j)^2}$, wobei $\Delta y_j$ die Differenz der Funktionswerte am Ende und am Anfang des jeweiligen Geradenstücks ist. Diese Größe ändert sich, wenn man von einen zum nächsten Geradenstück geht, was den Index $j$ erklärt. Die Länge des Graphens wird dann durch die summierte Länge aller Geradenstücke approximiert. Erhöhen wir nun wieder die Anzahl der Geradenstücke, bzw. verkleinern $\Delta x$, so nähern wir uns der tatsächlichen Länge immer weiter an. Im Limes infinitesimal kleiner Geradenstücke $\Delta s_j$ erhält man das **Kurvenintegral** entlang des Weges $\mathcal C$

\begin{equation}
    L = \sum_{j = 1}^{N} \Delta s_j \rightarrow  \int_{\mathcal{C}} \mathrm{d}s \ .
\end{equation}

Die Größe $\mathcal{C}$ bezeichnet den durch die Funktion $f(x)$ beschriebenen Weg.

### II. Kurvenintegrale von Kraftfeldern

Sehr häufig werden Kurvenintegrale auch dazu genutzt Kraftfelder entlang einer Kurve zu integrieren. Dies führt auf die Arbeit. Im Hinblick auf unseren Bergsteiger aus der vorherigen Aufgabe kann man so bspw. die Frage stellen, welche Arbeit er entlang seines Weges den Berg hinauf aufbringen muss. Die Arbeit $\Delta W$, die aufgebracht werden muss um ein Objekt im Kraftfeld $\vec{F}$ entlang eines kleinen Wegstücks $\Delta \vec{x}$ zu bewegen ist

\begin{equation}
    \Delta W = \vec{F} \cdot \Delta \vec{x} \ .
\end{equation}

Hierbei gehen wir davon aus, dass die Kraft entlang dieses Wegstückes als konstant angenommen werden kann, sich also $\vec F$ über $\Delta \vec x$ nur schwach verändert. Soll ein Objekt durch dieses Kraftfeld entlang eines Weges $\mathcal{C}$ bewegt werden, erhält man die dafür benötigte Arbeit durch das Integral

\begin{equation}
    W = \sum_{j = 1}^{N} \Delta W_j = \sum_{j = 1}^{N} \vec{F}(\vec{x}_j) \cdot \Delta \vec{x}_j \rightarrow \int_\mathcal{C} \vec{F}(\vec{x}) \cdot \mathrm{d}\vec{x}
\end{equation}

in Analogie zum obigen Kurvenintegral der Weglänge. Es ist wichtig zu beachten, dass das Kraftfeld nun - falls ortsabhängig - für jeden Summanden neu ausgewertet werden muss; in der Summe tritt $\vec{F}(\vec{x}_j)$ auf. Zur Berechnung des Integrals kann also wieder auf die Approximation des Weges als Reihe von Geradenstücken zurückgegriffen werden. Anstatt jedoch die Längen der Geradenstücke zu berechnen, berechnet man nun das Skalarprodukt zwischen der Kraft $\vec{F}(\vec{x}_j)$ und dem Vektor $\Delta\vec{x}_j$, der durch das Geradenstück gebildet wird.

Im Folgenden werden wir zwei verschiedene Kraftfelder betrachten:

\begin{equation}
    \vec{F}_G = -mg\begin{pmatrix} 0 \\ 1 \\ 0 \end{pmatrix} \quad \mathrm{und} \quad \vec{F}_R (\vec{x}) = -\frac{mg}{\sqrt{x^2 + y^2}} \begin{pmatrix} -y \\ x \\ 0 \end{pmatrix} \ .
\end{equation}

Das erste Kraftfeld entspricht dem homogenen Gravitationsfeld der Erde mit $g = 9.81 \ \mathrm{m} / \mathrm{s}^2 $ und das zweite einem "ausgedachten", ortsabhängigen Kraftfeld (ohne physikalische Bedeutung), welches an jeder Stelle den gleichem Betrag wie das Gravitationsfeld hat.
