Fast alles in unserer Welt wächst und verfällt. Dynamische Prozesse, ob physikalischer, chemischer, biologischer oder soziologischer Natur, können mit Hilfe mathematischer Modelle approximativ beschrieben werden. Als Hauptwerkzeuge für mathematische Modellierungen werden Differentialgleichungen (DGLn) verwendet, d.h. Gleichungen, die eine Funktion und eine oder mehrere ihrer Ableitungen enthalten. Die hier gesuchten Unbekannten sind jedoch keine Zahlenwerte, wie es bei gewöhnlichen Gleichungen z.B. bei einer quadratischen Gleichung
</div>
<p></p>
\begin{equation*}
 x^2 -8x + 5 = 0 \quad \text{mit Lösungen $x_1=7$ und $x_2=1$}~,
\label{eq:gew_DGL} \tag{1}
\end{equation*}    
<p></p>
<div style="text-align: justify">
der Fall ist, sondern Funktionen. In diesem Notebook werden wir uns auf <strong>gewöhnliche DGLn</strong> (eine Veränderliche) konzentrieren.
</div><br>
<div style="text-align: justify">
Eine DGL definiert die Beziehung zwischen einer unbekannten Funktion und ihren Ableitungen. Die erste Ableitung stellt hierbei die Änderungsrate der unbekannten Funktion z.B. einer physikalischen Größe, dar, die zweite Ableitung charakterisiert die Krümmung der Funktion usw. Eine der häufigsten Arten, wie eine Größe (z.B. mit der Zeit) wachsen oder abnehmen kann, ist die exponentielle Änderung. Betrachten wir das folgende einführende Beispiel einer DGL
</div>
<p></p>
\begin{equation*}
\cfrac{\mathrm{d}}{\mathrm{d} t} y(t) = \lambda y(t) \quad \text{mit} \quad \lambda \in \mathbb{R}~.
\label{eq:exp_change_DGL} \tag{2}
\end{equation*}
<p></p>
<div style="text-align: justify">
Falls der Parameter $\lambda$ positiv ist, wird dieser als Wachstumsrate oder im negativen Fall als Zerfallsrate bezeichnet. Die Lösung der DGL ergibt sich z.B. für $\lambda=1$ via $y(t)=\exp \big\{ t \big\}$ oder für beliebiges $\lambda$ via $y(t)=\exp \big\{ \lambda t \big\}$ und durch Anwendung der Kettenregel zu
</div>
<p></p>
\begin{equation*}
y(t) = C \exp \big\{ \lambda t \big\} \quad \text{mit} \quad C \in \mathbb{R}~.
\label{eq:exp_change_DGL_solution} \tag{3}
\end{equation*}
<p></p>
<div style="text-align: justify">
Aufgrund der beliebigen Integrationskonstanten $C$ existieren somit unendlich viele Lösungsfunktionen. Mit einem Anfangswert für $y(t)$ können wir $C$ bestimmen und eine bestimmte Lösung aus der unendlich großen Lösungsschar herausfiltern. Als Beispiel eines <strong>Anfangwertproblems (AWP)</strong> nehmen wir den radioaktiven Zerfall des Kohlenstoff-14-Isotopes mit der Zerfallsrate $\lambda_{C_{14}} = 1.22 \cdot 10^{-4} /\text{yr}$. Wir starten mit der folgenden DGL 
</div>
<p></p>
\begin{equation*}
\cfrac{\mathrm{d}}{\mathrm{d} t} n(t) = - \lambda_{C_{14}} n(t) \quad \text{und dem Anfangswert} \quad n(t=0~\mathrm{s}) = 100~.
\label{eq:decay_DGL} \tag{4}
\end{equation*}
<p></p>
Die Größe $n$ stellt hierbei die Anzahl der C-14-Isotope dar. Die Lösung kann mit Hilfe der Gl. (3) sofort abgelesen werden
<p></p>
\begin{equation*}
n(t) = 100 \exp \big\{ - \lambda_{C_{14}} t \big\} \quad \text{mit} \quad C= n(0) = 100~.
\label{eq:decay_DGL_solution} \tag{5}
\end{equation*}
<p></p>
Anschließend plotten wir die Lösungsfunktion $n(t)$ und ermitteln direkt aus dem Plot die Halbwertszeit $T_{1/2}$, d.h. die Zeitspanne, nach der die Anzahl $n$ der C-14-Isotope auf die Hälfte des Startwertes abgeklungen ist:

```python
import numpy as np
import matplotlib.pyplot as plt

def decay (t, rate, iniVal):
    return iniVal*np.exp(-rate*t)

n0 = 100
lambdaC14 = 1.22e-4
tIni = 0 # Startzeit
tFin = 25000 # Endzeit
fSam = 10 # Abstastfrequenz (Samplingfrequenz), hier 10 Abtastungen/Jahr
samplingPoints = fSam*tFin # Anzahl der abgetastetetn Punkte
tEval = np.linspace(tIni, tFin, samplingPoints, endpoint=True)
n = np.zeros_like(tEval)
n = decay(tEval, lambdaC14, n0)
Thalf = 0.0
for i in range(0, len(n)): # Brute-Force-Methode für die Bestimmung der Halbwertszeit.
    if (n[i] <= n0/2):
        nHalf = n[i]
        Thalf = i*tFin/samplingPoints
        print(r"nHalf = ", nHalf, "; Thalf = ", Thalf, "yr" )        
        break
fig, axs = plt.subplots(1, 1,  figsize=(6, 6))
axs.set_xlabel(r"$t$ (yr)") # Jahre
axs.set_ylabel(r"$n$")
axs.set_xlim([tIni, tFin])
axs.set_ylim([0,n0])
axs.locator_params(axis='x', nbins=10)
axs.locator_params(axis='y', nbins=10)
axs.grid()
axs.plot(tEval,n)
axs.axvline(Thalf,0, nHalf/n0, color="red", linestyle="--")
axs.axhline(nHalf,0, Thalf/tFin, color="red", linestyle="--")
plt.show()    
```

<div style="text-align: justify">
Die Halbwertszeit beträgt somit $T_{1/2} \approx 5682$ Jahre. Eine höhere Samplingfrequenz (Abtastrate) <code>fSam</code> verbessert die Genauigkeit, streckt jedoch die Berechnungszeit in die Länge.
</div>
</div>
<hr style="width:50%">
<div style="text-align: justify">
Bevor wir zu den einzelnen Aufgaben übergehen, führen wir ganz kurz einige wichtige Begriffe ein, um das hier zubearbeitende Thema besser einzuordnen zu können. Die DGL (2) liegt in der <strong>expliziten Darstellung</strong> vor, d.h. auf der linken Seite steht die Ableitung mit der höchsten <strong>Ordnung</strong> (hier Ordnung = 1) und auf der rechten Seite stehen die restlichen Terme (weitere Ableitungen der unbekannten Funktion $y(x)$ und Funktionen der Variablen $x$). Eine explizite DGL $3.$ Ordnung würde im Allgemeinen folgendermaßen aussehen
</div>
<p></p>
\begin{equation*}
y^{(3)}(x) = f \Big(x,y(x),y^{\prime}(x),y^{\prime \prime}(x)\Big) \quad \text{mit} \quad \cfrac{\mathrm{d}}{\mathrm{d} x} y(x) = y^{\prime} (x) \quad \text{und} \quad \cfrac{\mathrm{d}^3}{\mathrm{d} x^3} y(x) = y^{\prime \prime \prime} (x) = y^{(3)} (x)~.
\label{eq:explizite_DGL} \tag{6}
\end{equation*}
<p></p>
<div style="text-align: justify">
Die rechte Seite der Gleichung könnte z.B. die folgende Darstellung haben 
<p></p>
\begin{equation*}
f \Big(x,y(x),y^{\prime}(x),y^{\prime \prime}(x)\Big) = \frac{1}{4}y^{\prime \prime}(x) +2 y^{\prime}(x) - y^2(x) + \sin(x)~.
\label{eq:rs_explizite_DGL} \tag{7}
\end{equation*}
<p></p>
</div>    
<div style="text-align: justify">
Lässt sich die DGL nicht nach $y^{(3)}(x)$ auflösen, so muss diese in der <strong>impliziten Darstellung</strong>, d.h.
</div>
<p></p>
\begin{equation*}
F \Big(x,y(x),y^{\prime}(x),y^{\prime \prime}(x),y^{(3)}(x) \Big) = 0
\label{eq:implizite_DGL} \tag{8}
\end{equation*}
<p></p>
<div style="text-align: justify">    
gelöst werden. Diese Art von DGLn werden wir in diesem Notebook nicht behandeln.<br> 
Im Falle eines AWP müssen wir für die oben genannte DGL 3. Ordnung mehrere Anfangswerte festlegen. Um eine eindeutige Lösung zu erhalten, werden an einer Stelle $x_0$ soviele Bedingungen an die Funktion und ihre Ableitungen gestellt, wie die Ordnung der Gleichung ist. Für eine DGL 3. Ordnung benötigen wir somit drei Anfangswerte
</div> 
<p></p>
\begin{align*}    
    y(x_0) &= y_0 \\
    \cfrac{\mathrm{d}}{\mathrm{d}x} y(x) \Bigg|_{x=x_0} &= y_1 \\
    \cfrac{\mathrm{d}^{2}}{\mathrm{d}x^{2}} y(x) \Bigg|_{x=x_0} &= y_{2}~. 
    \label{eq:Anfangswerte} \tag{9}
\end{align*}
<p></p>
<div style="text-align: justify">    
   Des Weiteren lassen sich gewöhnliche DGLn in <strong>lineare</strong> und <strong>nicht-lineare DGLn</strong> klassifizieren. Zum Beispiel ist,
</div>
<ul>
<li><div style="text-align: justify">$x^2 y^{\prime \prime}(x) = y(x)$<br>
ist eine lineare DGL zweiter Ordnung (Funktion $y(x)$ und deren Ableitungen nur in erster Potenz, nicht miteinander multipliziert), </div>  
</li>   
<li><div style="text-align: justify">$y(x) y^{\prime}(x) = x$<br>
ist hingegen eine nicht-lineare DGL erster Ordnung.
</div>
</li>
</ul>
<div style="text-align: justify">
Eine lineare DGL 3. Ordnung  würde im Allgemeinen die folgende Darstellung haben   
</div>
<p></p>
\begin{align*}    
y^{\prime \prime \prime}(x) &= a_0(x)~ y(x) + a_1(x)~ y^{\prime}(x) + a_2(x)~ y^{\prime \prime}(x) + g(x) \\[1em] &= a_0(x)~ y^{(0)}(x) + a_1(x)~ y^{(1)}(x) + a_2(x)~ y^{(2)}(x) + g(x)\\[1em] &= \sum \limits_{k=0}^{2} a_k(x)~ y^{(k)}(x) + g(x) \quad \text{mit} \quad g(x) \hat{=}~
\text{beliebige Funktion von $x$}~.
\label{eq:lin_dgl_3ord} \tag{10}
\end{align*}    
<p></p>
<div style="text-align: justify">        
Ist $g(x) = 0$, so heißt die lineare DGL <strong>homogen</strong>, ansonsten für $g(x) \neq 0$ <strong>inhomogen</strong>. Die oben besprochene DGL (2) ist somit eine lineare homogene DGL 1. Ordnung (auch autonome DGL genannt, da sie nicht explizit von $x$ abhängt).
</div> 
<p></p>
<div style="text-align: justify">
In diesem Notebook werden wir uns auf explizite DGLn erster Ordnung der Form
</div>
<p></p>   
\begin{equation*}
y^{\prime}(x) = f \big( x,y(x) \big)
\label{eq:explizite_DGL_1o} \tag{11}
\end{equation*}
<p></p>
<div style="text-align: justify">    
konzentrieren und deren Lösungen mit Hilfe von <strong>Richtungsfeldern</strong> visualisieren.
</div>    

<a id="Richtungsfeld"></a>
<div style=" border: 2px solid black; margin: 10px auto; padding: 10px; border-radius: 10px; background-color:  #f4ecf7 ;">
<b>Richtungsfeld:</b> 
<ul>
<div style="text-align: justify">
Das Richtungsfeld (RF) einer gewöhnlichen DGL $\cfrac{\mathrm{d}}{\mathrm{d} x} y(x) = f \big( x,y(x) \big)$ lässt sich einfach mit Hilfe eines 2D-Grids (Gitters) erzeugen. Hierbei wird an jeden Punkt $P = \big( x,y(x) \big)$ des 2D-Grids ein Vektor $\vec{t} = \Big( 1, f \big(x,y(x) \big) \Big)^T$ angehängt. Die Richtung dieses Vektors fällt mit der am Punkt $P$ anliegenden Tangente der Funktion $y(x)$ zusammen. Die Tangentensteigung $m = \cfrac{\Delta y}{\Delta x} = \cfrac{f \big( x,y(x) \big)}{1}$ entspricht genau der Ableitung $y^{\prime} (P)$. Die Lösungskurve $y(x)$ ist somit in jedem Grid-Punkt tangential zum Richtungsfeld. Verfolgt man die Richtungsvektoren entlang eines Pfades, so lässt sich ein grober Verlauf der Lösungskurve konstruieren (einzeichnen). Mit Hilfe des RF ist es somit möglich, Lösungskurven für Anfangswertprobleme grafisch zu visualisieren, ohne dabei die tatsächliche Lösung berechnet zu haben. Es ist üblich die Vektoren des RF ohne Pfeilspitzen zu zeichnen, was darauf hinweist, dass sie in beide Richtungen verfolgt werden können.
</div>
</ul>

</div>   

Um das RF zu konstruieren, müssen wir, wie schon oben erwähnt, nicht die DGL lösen. Für die numerische Auswertung und Visualisierung der Ergebnisse benötigen wir die folgenden Python-Funktionen:
</div>    
<ul>
<li>
<code>numpy.meshgrid</code> (<a href="https://numpy.org/doc/stable/reference/generated/numpy.meshgrid.html">meshgrid-Doc</a>) $\longrightarrow$ erstellt ein (2D)-Grid,
</li>
<li>
<code>pyplot.quiver</code> (<a href="https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.quiver.html">quiver-Doc</a> und <a href="https://matplotlib.org/stable/gallery/images_contours_and_fields/quiver_demo.html#sphx-glr-gallery-images-contours-and-fields-quiver-demo-py">quiver-Besipiele</a>) $\longrightarrow$ stellt ein RF auf dem 2D-Grid dar,
</li>
</ul>    

<a id="2D_Grid"></a>
<div style=" border: 2px solid green;margin: 10px auto; padding: 10px; border-radius: 10px; background-color: beige;">
<div style="text-align: justify">
Beispiele eines 2D-Grids:
</div>
</div>

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameter
xLim = [-2,2]
yLim = [-2,2]
resolution = 5
# Initialisiere benötigte Arrays mit 0
X1 = np.zeros(resolution)
X2 = np.zeros(resolution)
Y1 = np.zeros(resolution)
Y2 = np.zeros(resolution)
Xgrid1 = np.zeros((resolution,resolution))
Xgrid2 = np.zeros((resolution,resolution))
Ygrid1 = np.zeros((resolution,resolution))
Ygrid2 = np.zeros((resolution,resolution))
###
X1 = np.linspace(xLim[0],xLim[1],resolution, endpoint=True)
Y1 = np.linspace(yLim[0],yLim[1],resolution, endpoint=True)
X2 = np.linspace(xLim[0],xLim[1],resolution, endpoint=False)
Y2 = np.linspace(yLim[0],yLim[1],resolution, endpoint=False)
Xgrid1, Ygrid1 = np.meshgrid(X1, Y1)
Xgrid2, Ygrid2 = np.meshgrid(X2, Y2)
###
fig, axs = plt.subplots(1, 2,  figsize=(14, 7))
axs[0].axvline(0, c="black")
axs[0].axhline(0, c="black")
axs[0].text(xLim[1]+1.1, 0, r"x")
axs[0].text(0, yLim[1]+1.1, r"y")
axs[0].set_xlim([xLim[0]-1, xLim[1]+1])
axs[0].set_ylim([yLim[0]-1, yLim[1]+1])
#axs[0].locator_params(axis='x', nbins=6)
#axs[0].locator_params(axis='y', nbins=6)
axs[0].grid()
axs[0].plot(Xgrid1, Ygrid1, marker="o", color="red", linestyle="none")
###
axs[1].axvline(0, c="black")
axs[1].axhline(0, c="black")
axs[1].text(3.1, 0, r"x")
axs[1].text(0, 3.1, r"y")
axs[1].set_xlim([xLim[0]-1, xLim[1]+1])
axs[1].set_ylim([yLim[0]-1, yLim[1]+1])
#axs[1].locator_params(axis='x', nbins=6)
#axs[1].locator_params(axis='y', nbins=6)
axs[1].grid()
axs[1].plot(Xgrid2, Ygrid2, marker="o", color="blue", linestyle="none")
#plt.tight_layout()
plt.show()
```

<a id="Quiver_Plot"></a>
<div style=" border: 2px solid green;margin: 10px auto; padding: 10px; border-radius: 10px; background-color: beige;">
<div style="text-align: justify">
Einfache Beispiele von Quiver-Plots:

```python
import numpy as np
import matplotlib.pyplot as plt

xLimQ1 = [1,10]
yLimQ1 = [1,10]
resolutionQ1 = 10
XQ1 = np.zeros(resolutionQ1)
YQ1 = np.zeros(resolutionQ1)
XgridQ1 = np.zeros((resolutionQ1,resolutionQ1))
YgridQ1 = np.zeros((resolutionQ1,resolutionQ1))
xLimQ2 = [1,20]
yLimQ2 = [1,20]
resolutionQ2 = 20
XQ2 = np.zeros(resolutionQ2)
YQ2 = np.zeros(resolutionQ2)
XgridQ2 = np.zeros((resolutionQ2,resolutionQ2))
YgridQ2 = np.zeros((resolutionQ2,resolutionQ2))
#
# 2D-Grid und das zugehörige Richtungsfeld
XQ1 = np.linspace(xLimQ1[0],xLimQ1[1],resolutionQ1, endpoint=True)
YQ1 = np.linspace(yLimQ1[0],yLimQ1[1],resolutionQ1, endpoint=True)
XgridQ1, YgridQ1 = np.meshgrid(XQ1, YQ1)
XgridQnorm1 = XgridQ1/np.sqrt(XgridQ1**2 + YgridQ1**2) # die Vektoren wurden hier auf die Länge 1 normiert
YgridQnorm1 = YgridQ1/np.sqrt(XgridQ1**2 + YgridQ1**2) # mit zusätzlichern Quiver-Plot-Parametern kann man die Länge weiter skalieren (siehe unten) 
XQ2 = np.linspace(xLimQ2[0],xLimQ2[1],resolutionQ2, endpoint=True)
YQ2 = np.linspace(yLimQ2[0],yLimQ2[1],resolutionQ2, endpoint=True)
XgridQ2, YgridQ2 = np.meshgrid(XQ2, YQ2)
#
fig, axs = plt.subplots(1, 3,  figsize=(21, 7))
axs[0].set_title("RF, autom. Skalierung der Vektorlänge \n verhindert das Überlappen der Vektoren")
axs[0].set_xlabel(r"$x$")
axs[0].set_ylabel(r"$y$")
axs[0].set_xlim([xLimQ1[0], xLimQ1[1]+1])
axs[0].set_ylim([yLimQ1[0], yLimQ1[1]+1])
axs[0].locator_params(axis='x', nbins=10)
axs[0].locator_params(axis='y', nbins=10)
axs[0].grid()
axs[0].plot(XgridQ1, YgridQ1, marker="o", color="grey", linestyle="none")
axs[0].quiver(XQ1,YQ1,XgridQ1,YgridQ1, width=0.006, color="red", label=r"")
#
axs[1].set_title("gleiches RF mit autom. Skalierung der Vektorlänge, \n doppelt großes Gebiet")
axs[1].set_xlabel(r"$x$")
axs[1].set_ylabel(r"$y$")
axs[1].set_xlim([xLimQ2[0], xLimQ2[1]+1])
axs[1].set_ylim([yLimQ2[0], yLimQ2[1]+1])
axs[1].locator_params(axis='x', nbins=10)
axs[1].locator_params(axis='y', nbins=10)
axs[1].grid()
axs[1].plot(XgridQ2, YgridQ2, marker="o", color="grey", linestyle="none")
axs[1].quiver(XQ2,YQ2,XgridQ2,YgridQ2, width=0.006, color="red", label=r"")
#
axs[2].set_title("gleiches RF, jedoch mit normierten Richtungsvektoren \n(keine autom. Skalierung, keine Pfeilspitze)")
axs[2].set_xlabel(r"$x$")
axs[2].set_ylabel(r"$y$")
axs[2].set_xlim([xLimQ1[0], xLimQ1[1]])
axs[2].set_ylim([yLimQ1[0], yLimQ1[1]])
axs[2].locator_params(axis='x', nbins=10)
axs[2].locator_params(axis='y', nbins=10)
axs[2].grid()
axs[2].plot(XgridQ1, YgridQ1, marker="o", color="grey", linestyle="none")
axs[2].quiver(XQ1,YQ1,XgridQnorm1,YgridQnorm1, angles='xy', scale_units='xy', scale=1, width=0.006, headaxislength=0, headwidth=0, headlength=0, color="red", label=r"") 
plt.show()
```

Die Quiver-Plot-Funktion skaliert automatisch den Betrag der RF-Vektoren und verhindert somit, dass diese überlappen. Die angezeigten Beträge der Vektoren entsprechen jedoch nicht der tatsächlichen "Stärke des Feldes". Das Verhältnis der einzelnen Beträge zueinander wird jedoch auf dem ganzen Grid korrekt angezeigt. Die automatische Skalierung kann man außerdem mit den folgenden Parametern beeinflussen: <code>scale</code>, <code>scale_units</code> und <code>angles</code>. Die Darstellung der Vektorspitze kann mit <code>headlength</code>, <code>headaxislength</code> und <code>headwidth</code> modifiziert werden (siehe auch Quiver-Doc).

<br/>    
<b>Teilaufgabe a)</b>
<div style="text-align: justify">    
Erzeugen Sie zuerst ein 2D-Grid, welches aus 81 Punkten (grau) besteht und innerhalb der Koordinatengrenzen $(x,y) \in \big[-2,2 \big]$ liegt.<br/>

Plotten Sie nun mit Hilfe der Grid-Punkte das zugehörige normierte RF (rot, keine Pfeilspitzen). Verwenden Sie für den Quiver-Plot <code>scale=4</code>.

```python
import sys
import numpy as np
import matplotlib.pyplot as plt

xLimEx1 = [-2,2]
yLimEx1 = [-2,2]
resolutionEx1 = 9
XEx1 = np.zeros(resolutionEx1)
YEx1 = np.zeros(XEx1.shape)
YisoclinesEx1 = np.zeros((7,resolutionEx1))
XgridEx1 = np.zeros((resolutionEx1,resolutionEx1))
YgridEx1 = np.zeros(XgridEx1.shape)
dyEx1 = np.zeros(XgridEx1.shape)
#
XEx1 = np.linspace(xLimEx1[0],xLimEx1[1],resolutionEx1)
YEx1 = np.linspace(yLimEx1[0],yLimEx1[1],resolutionEx1)
XgridEx1, YgridEx1 = np.meshgrid(XEx1, YEx1)

dyEx1 = YgridEx1 - XgridEx1
dxEx1 = np.ones(dyEx1.shape)
dxEx1norm = dxEx1/np.sqrt(dxEx1**2 + dyEx1**2)
dyEx1norm = dyEx1/np.sqrt(dxEx1**2 + dyEx1**2)
#
fig, axs = plt.subplots(1, 1,  figsize=(8, 8))
axs.set_title(r"normiertes Richtungsfeld")
axs.axvline(0, c="black")
axs.axhline(0, c="black")
axs.set_xlabel(r"$x$")
axs.set_ylabel(r"$y$")
axs.set_xlim([xLimEx1[0], xLimEx1[1]])
axs.set_ylim([yLimEx1[0], yLimEx1[1]])
axs.grid()

axs.plot(XgridEx1, YgridEx1, marker='o', color='grey', linestyle='none')
axs.quiver(XEx1,YEx1,dxEx1norm,dyEx1norm, angles='xy', scale_units='xy', scale=4, width=0.006, headaxislength=0, headwidth=0, headlength=0, color="red", label=r"")  
plt.show()
```

Verwenden Sie den in Aufgabe 1 erzeugten Plot, samt RF.

```python
import numpy as np
from scipy.integrate import odeint

iniValues1=[0.1,-1,-1.2]

def yPrime(y,x):
    dydx = y-x
    return dydx

xEval = np.linspace(-2, 2, 40)
solutionNum2 = odeint(yPrime, iniValues1, xEval)

fig, axs = plt.subplots(1, 1,  figsize=(8, 8))
axs.set_title(r"normiertes Richtungsfeld mit drei Lösungen")
axs.axvline(0, c="black")
axs.axhline(0, c="black")
axs.set_xlabel(r"$x$")
axs.set_ylabel(r"$y$")
axs.set_xlim([xLimEx1[0], xLimEx1[1]])
axs.set_ylim([yLimEx1[0], yLimEx1[1]])
axs.grid()
axs.plot(XgridEx1, YgridEx1, marker='o', color='grey', linestyle='none')
axs.quiver(XEx1,YEx1,dxEx1norm,dyEx1norm, angles='xy', scale_units='xy', scale=4, width=0.006, headaxislength=0, headwidth=0, headlength=0, color="red", label=r"")
#axs.quiver(XEx1,YEx1,dxEx1,dyEx1, width=0.004, color="red", label=r"")
axs.plot(xEval, solutionNum2, "--", linewidth=3.0)
plt.show()
```

<ul>
<li>$C$: Integrationskonstane(n),</li>
<li>$x$: unabhängige Variable.</li>
</ul>    
Plotten Sie die Lösungsfunktion in Form einer Lösungsschar für 40 verschiedene äquidistante Werte von $C$ für $x \in \big[ -3,6 \big]$. Verwenden Sie den in Aufgabe 1 erzeugten Plot, samt RF.

```python
def solFunc(C, x):
    return C*np.exp(x)+x+1

xEval = np.linspace(-2, 2, 40)
Ceval = np.linspace(-3, 6, 40)

fig, axs = plt.subplots(1, 1,  figsize=(8, 8))
axs.set_title(r"normiertes Richtungsfeld mit Lösungsschar")
axs.axvline(0, c="black")
axs.axhline(0, c="black")
axs.set_xlabel(r"$x$")
axs.set_ylabel(r"$y$")
axs.set_xlim([xLimEx1[0], xLimEx1[1]])
axs.set_ylim([yLimEx1[0], yLimEx1[1]])
axs.grid()
axs.plot(XgridEx1, YgridEx1, marker='o', color='grey', linestyle='none')
axs.quiver(XEx1,YEx1,dxEx1norm,dyEx1norm, angles='xy', scale_units='xy', scale=4, width=0.006, headaxislength=0, headwidth=0, headlength=0, color="red", label=r"")
for i in Ceval:
    axs.plot(xEval, solFunc(i, xEval), ":", linewidth=3.0, color="blue")    
#
plt.show()
```
